```python
img = cv2.imread('minion.jpg')
B, G, R = cv2.split(img)

cv2_imshow(B)
cv2_imshow(G)
cv2_imshow(R)
```

```python
from matplotlib import cm
#subplot do azul (B)
plt.subplot(1,3,1)
plt.imshow(B, cmap='gray')
plt.axis('off')
plt.title('Azul - B')
#subplot do verde (G)
plt.subplot(1,3,2)
plt.imshow(G, cmap='gray')
plt.axis('off')
plt.title('Verde - G')
#subplot do vermelho (R)
plt.subplot(1,3,3)
plt.imshow(R, cmap='gray')
plt.axis('off')
plt.title('Vermelho - R')
```
