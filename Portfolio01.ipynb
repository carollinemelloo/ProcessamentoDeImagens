```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
img = cv2.imread('jovenstitans.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_gray
```

```python
img_1bit = (img_gray // 128) * 255
img_2bits = (img_gray // 64) * 85
img_4bits = (img_gray // 16) * 17
img_8bits = (img_gray // 1) * 1
```

```python
fig, ax = plt.subplots(2, 3, figsize=(12, 8))

ax[0, 0].imshow(img_gray, cmap='gray')
ax[0, 0].set_title("Original (8 bits - 256 níveis)")
ax[0, 0].axis('off')

ax[0, 1].imshow(img_8bits, cmap='gray')
ax[0, 1].set_title("Quantizada: 8 bits (256 níveis)")
ax[0, 1].axis('off')

ax[0, 2].imshow(img_4bits, cmap='gray')
ax[0, 2].set_title("Quantizada: 4 bits (16 níveis)")
ax[0, 2].axis('off')

ax[1, 0].imshow(img_2bits, cmap='gray')
ax[1, 0].set_title("Quantizada: 2 bits (4 níveis)")
ax[1, 0].axis('off')

ax[1, 1].imshow(img_1bit, cmap='gray')
ax[1, 1].set_title("Quantizada: 1 bit (2 níveis)")
ax[1, 1].axis('off')

ax[1, 2].axis('off')

plt.tight_layout()
plt.savefig('quantizacao.png')
plt.show()
```


# O que acontece com os detalhes e com as transições de intensidade da imagem à medida que reduzimos o número de bits?

R = Quando os bits são reduzidos, a imagem vai ganhando constraste. Consequentemente, a imagem vai ficando menos detalhada pois o que tem menos contraste deixa de ficar visível. até que na imagem de 1 bit ficam apenas regiões pretas e brancas.
